```python
# Reload the dataset for Task 2 (fresh copy with all columns)
housing_url = "https://raw.githubusercontent.com/ageron/handson-ml/master/datasets/housing/housing.csv"
df = pd.read_csv(housing_url)

print(f"Dataset reloaded: {df.shape[0]} rows × {df.shape[1]} columns")
print(f"\nColumns available: {list(df.columns)}")
print(f"\nFirst few rows:")
print(df.head())
```

```python
# Categorize housing age into groups
df['age_category'] = df['housing_median_age'].apply(
    lambda age: 'New' if age <= 10 
    else 'Moderate' if age <= 30 
    else 'Old'
)
```

```python
# Categorize income levels
df['income_category'] = df['median_income'].apply(
    lambda income: 'Low' if income < 3 
    else 'Medium' if income < 6 
    else 'High'
)
```

```python
# Define mapping dictionary
ocean_mapping = {
    'NEAR BAY': 'Coastal',
    'NEAR OCEAN': 'Coastal',
    '<1H OCEAN': 'Coastal',
    'INLAND': 'Inland',
    'ISLAND': 'Island'
}

# Apply mapping
df['location_type'] = df['ocean_proximity'].map(ocean_mapping)
```

```python
# Clean string column - standardize ocean_proximity
df['ocean_proximity_clean'] = df['ocean_proximity'].apply(
    lambda x: x.strip().lower() if isinstance(x, str) else x
)

print("Task 2: String cleaning completed")
print(df['ocean_proximity_clean'].unique())
```

```python
# Display all Task 2 transformations
print("=" * 80)
print("TASK 2: DATA TRANSFORMATION RESULTS")
print("=" * 80)

print("\n1. Apply & Lambda - Age Categorization:")
print(df[['housing_median_age', 'age_category']].head(10))

print("\n2. Apply & Lambda - Income Categorization:")
print(df[['median_income', 'income_category']].head(10))

print("\n3. Map - Ocean Proximity Transformation:")
print(df[['ocean_proximity', 'location_type']].head(10))

print("\n4. Lambda - String Cleaning:")
print(df[['ocean_proximity', 'ocean_proximity_clean']].head(10))

print("\n" + "=" * 80)
print("SUMMARY OF NEW COLUMNS CREATED:")
print("=" * 80)
print(f"Total new columns added: 4")
print(f"- age_category (from housing_median_age)")
print(f"- income_category (from median_income)")
print(f"- location_type (from ocean_proximity)")
print(f"- ocean_proximity_clean (from ocean_proximity)")
print("=" * 80)
```
